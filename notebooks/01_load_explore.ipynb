## 📓 Notebook 01: Load & Explore

**Skills practiced:** Loading a CSV, converting a Yes/No target to 0/1, checking class balance.

**Goal:** Load the churn dataset and record its churn rate, which you'll need to interpret model scores later.

```python
import pandas as pd
```

```python
df=pd.read_csv('../data/raw/Telco_customer_churn.csv').drop(columns=['customerID'])
```

```python
df.info()
```

```python
df=df.dropna(subset=['Churn'])
```

```python
df.Churn=df.Churn.map({'Yes':1, 'No':0})
df.Churn.value_counts(normalize=True)
```

```python
df.to_csv('../data/processed/Telco_customer_churn.csv', index=False) 
```
